## Cell 2: 2 — SindhiNormalizer

````python
from sindhinltk.normalizer import SindhiNormalizer
norm = SindhiNormalizer()

examples = [
    ("  سنڌي ٻولي  ",               False),
    ("سنڌي    ٻولي	آهي",           False),
    ("ھوُ ھر روزَ اِسڪول وَڃي ٿو",  True),
    ("پاڪستان ۾ چئني صوبن ۾ سنڌي ڳالهائي وڃي ٿي",  False)
]

for text, strip_diac in examples:
    result = norm.normalize(text, remove_diacritics=strip_diac)
    print(f"Input:  {repr(text)}")
    print(f"Output: {repr(result)}  (remove_diacritics={strip_diac})")
    print()
````

## Cell 3: 3 — SindhiTokenizer

````python
from sindhinltk.tokenizer import SindhiTokenizer
tok = SindhiTokenizer()

sentences = [
    "ڪاوڙيندڙ ماڻهو گھر ۾ مسئلا پيدا ڪندو آهي",
    "سنڌي ٻولي تمام مٺي ۽ خوبصورت آهي",
    "گھر جھيل دھرتي کھانو",
]

for s in sentences:
    tokens = tok.tokenize(s)
    fertility = len(tokens) / len(s.split())
    print(f"Input:     {s}")
    print(f"Tokens:    {tokens}")
    print(f"Fertility: {fertility:.2f}")
    print()

sents = tok.sent_tokenize("هو گھر ۾ آهي۔ سنڌي ٻولي سٺي آهي۔ توهان ڪيئن آهيو؟")
print(f"sent_tokenize output: {sents}")
print(f"Sentence count      : {len(sents)}")
````

---
## Cell 4: 4 — SindhiStemmer

````python
from sindhinltk.stemmer import SindhiStemmer
stm = SindhiStemmer()

cases = [
    ("ڪاوڙيندڙ",  "ڪاوڙ",   "participial active  -indur"),
    ("ڪندو",       "ڪ",      "progressive masc.   -ndo  "),
    ("وڃندا",      "وڃ",     "future              -nda  "),
    ("ڪيائين",     "ڪ",      "past continuous     -yain "),
    ("هلندي",      "هل",     "progressive fem.    -ndi  "),
    ("ڪندڙ",       "ڪ",      "participial         -ndr  "),
    ("سنڌي",       "سنڌي",   "no suffix — returned intact"),
]

print(f"{'Input':<16}  {'Result':<10}  Suffix Category")
print("─" * 55)
for word, expected, category in cases:
    result = stm.stem(word)
    ok = "✅" if result == expected else "❌"
    print(f"{ok}  {word:<14}  →  {result:<10}  {category}")

batch = ["ڪاوڙيندڙ", "ماڻهو", "ڪندو"]
print(f"\nstem_tokens input:  {batch}")
print(f"stem_tokens output: {stm.stem_tokens(batch)}")
````

## Cell 5: 5 — SindhiStopwords

````python
# SindhiStopwords
from sindhinltk.stopwords import SindhiStopwords
from sindhinltk.datasets import SindhiDatasets
sw = SindhiStopwords()
ds = SindhiDatasets()

# Loading list from datasets to show count
stopwords_list = ds.load_stopwords()
print(f"Total stopwords: {len(stopwords_list)}")
print()

checks = [
    ("آهي",         True),
    ("۽",           True),
    ("جي",          True),
    ("کي",          True),
    ("سنڌي",        False),
    ("ڪاوڙيندڙ",   False),
    ("گھر",         False),
]

print(f"{'Word':<18}  is_stopword  Classification")
print("─" * 50)
for word, expected in checks:
    result = sw.is_stopword(word)
    label  = "function word ✅" if result else "content word  ✅"
    print(f"  {word:<16}  {str(result):<11}  {label}")

print()
tokens   = ["سنڌي", "ٻولي", "آهي", "تمام", "سٺي"]
filtered = sw.remove_stopwords(tokens)
print(f"Input tokens :  {tokens}")
print(f"After filter :  {filtered}")
print(f"Removed      :  {list(set(tokens) - set(filtered))}")
````

---
## Cell 6: 6 — SindhiSentiment

````python
from sindhinltk.sentiment import SindhiSentiment
sa = SindhiSentiment()

test_cases = [
    ("سنڌي ٻولي تمام سٺي آهي",       "مثبت"),
    ("هو تمام خراب ماڻهو آهي",       "منفي"),
    ("هو گھر ۾ آهي",                 "غير جانبدار"),
    ("ڪاميابي خوشي آڻي ٿي",          "مثبت"),
    ("ناڪامي ۽ غم ڏکيو آهي",         "منفي"),
]

print(f"{'Text':<40}  {'Label':<16}  Score")
print("─" * 68)
for text, expected in test_cases:
    label = sa.analyze(text)
    score = sa.score(text)
    ok    = "✅" if label == expected else "❌"
    print(f"{ok}  {text:<38}  {label:<16}  {score}")

print()
print("analyze_detail output:")
detail = sa.analyze_detail("سنڌي ٻولي تمام سٺي آهي")
for key, val in detail.items():
    print(f"  {key:<12} : {val}")
````

## Cell 8: 8 — Full Pipeline (SindhiNLP)

````python
from sindhinltk.tokenizer import SindhiTokenizer
from sindhinltk.stemmer import SindhiStemmer
from sindhinltk.sentiment import SindhiSentiment

tok = SindhiTokenizer()
stm = SindhiStemmer()
sa = SindhiSentiment()

test_texts = [
    "سنڌي ٻولي تمام سٺي آهي",
    "ڪاوڙيندڙ ماڻهو گھر ۾ مسئلا پيدا ڪندو آهي",
    "شاهه لطيف جي شاعري سنڌ جو ادبي ورثو آهي",
]

for text in test_texts:
    tokens = tok.tokenize(text)
    stems = stm.stem_tokens(tokens)
    sentiment = sa.analyze(text)

    print(f"Input     :  {text}")
    print(f"Tokens    :  {tokens}")
    print(f"Stems     :  {stems}")
    print(f"Sentiment :  {sentiment}")
    print()
````

## Cell 9: 9 — Test Suite Summary

````python
#Test Suite Summary

from sindhinltk.tokenizer import SindhiTokenizer
from sindhinltk.normalizer import SindhiNormalizer
from sindhinltk.stemmer    import SindhiStemmer
from sindhinltk.stopwords  import SindhiStopwords
from sindhinltk.sentiment  import SindhiSentiment
from sindhinltk.datasets   import SindhiDatasets

passed = 0

tok  = SindhiTokenizer()
norm = SindhiNormalizer()
stm  = SindhiStemmer()
sw   = SindhiStopwords()
sa   = SindhiSentiment()
ds   = SindhiDatasets()

# Tokenizer (4 checks)
assert len(tok.tokenize("ڪاوڙيندڙ ماڻهو گھر ۾ مسئلا پيدا ڪندو آهي")) == 8;  passed += 1
assert tok.tokenize("") == [];                                                  passed += 1
assert tok.tokenize("سنڌي") == ["سنڌي"];                                        passed += 1
assert len(tok.sent_tokenize("آهي۔ ٻيو۔ ٽيون؟")) == 3;                         passed += 1

# Normalizer (3 checks)
assert norm.normalize("  سنڌي ٻولي  ") == "سنڌي ٻولي";                         passed += 1
assert "و" in norm.normalize("وَڃي", remove_diacritics=False);                  passed += 1
assert norm.normalize("") == "";                                                passed += 1

# Stemmer (7 checks)
for word, expected in [("ڪاوڙيندڙ","ڪاوڙ"),("ڪندو","ڪ"),("وڃندا","وڃ"),
                        ("ڪيائين","ڪ"),("هلندي","هل"),("ڪندڙ","ڪ"),("سنڌي","سنڌي")]:
    assert stm.stem(word) == expected; passed += 1

# Stopwords (3 checks)
# load_stopwords returns a dict; we sum the values to get the total count
sw_dict = ds.load_stopwords()
total_sw = sum(len(v) for v in sw_dict.values())
assert total_sw > 0;                         passed += 1
assert sw.is_stopword("آهي")  == True;       passed += 1
assert sw.is_stopword("سنڌي") == False;      passed += 1

# Sentiment (3 checks)
assert sa.analyze("سنڌي ٻولي تمام سٺي آهي") == "مثبت";        passed += 1
assert sa.analyze("هو تمام خراب ماڻهو آهي") == "منفي";         passed += 1
assert sa.analyze("هو گھر ۾ آهي")           == "غير جانبدار";  passed += 1

print("=" * 48)
print(f"  sindhinltk v1.3.1 — Test Suite")
print("=" * 48)
print(f"  Tokenizer  : 4  checks  ✅")
print(f"  Normalizer : 3  checks  ✅")
print(f"  Stemmer    : 7  checks  ✅")
print(f"  Stopwords  : 3  checks  ✅")
print(f"  Sentiment  : 3  checks  ✅")
print("─" * 48)
print(f"  RESULT: {passed} passed  |  0 failed  |  {passed} total")
print("=" * 48)
print("  ✅  ALL TESTS PASSED")
print("=" * 48)
````
